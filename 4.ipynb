```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
import numpy as np
df = pd.DataFrame({
    "Marks" : [80,70,None,85,60],
    "Attendence" : [90,None,86,80,75],
    "Department" : ["CSE","ECE","CSE",None,"ECE"]
})

#Mean Imputation
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

#Median Imputation
df["Attendence"] = df["Attendence"].fillna(df["Attendence"].median())

#Forward fill
df["Department"] = df["Department"].ffill()

#Fill missing categorical value using mode
df["Department"] = df["Department"].fillna(df["Department"].mode()[0])

#Label Encoding
encoder = LabelEncoder()
df["Department"] = encoder.fit_transform(df["Department"])
df
```
